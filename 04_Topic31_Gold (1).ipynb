```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

MY_ID = "tanmaya"

CATALOG = "workspace"
SCHEMA = f"capstone_{MY_ID}"

SILVER_CALLS = f"{CATALOG}.{SCHEMA}.silver_calls"
GOLD_TABLE = f"{CATALOG}.{SCHEMA}.gold_tower_hourly"

print("Silver table:", SILVER_CALLS)
print("Gold table:", GOLD_TABLE)
```

```python
silver = spark.table(SILVER_CALLS)

print("Silver row count:", silver.count())

display(silver.limit(10))
```

```python
calls_hourly = (
    silver
    .withColumn("call_date", F.to_date("start_ts_timestamp"))
    .withColumn("hour_of_day", F.hour("start_ts_timestamp"))
    .groupBy(
        "tower_id",
        "site_name",
        "city",
        "district",
        "capacity_channels",
        "call_date",
        "hour_of_day"
    )
    .agg(
        F.count("*").alias("calls"),
        F.sum(
            F.when(F.col("is_dropped") == True, 1).otherwise(0)
        ).alias("dropped_calls"),
        F.avg("duration_seconds_int").alias("avg_duration_seconds")
    )
)

display(calls_hourly.limit(20))
```

```python
calls_hourly = (
    silver
    .withColumn("call_date", F.to_date("start_ts_timestamp"))
    .withColumn("hour_of_day", F.hour("start_ts_timestamp"))
    .groupBy(
        "tower_id",
        "site_name",
        "city",
        "district",
        "capacity_channels",
        "call_date",
        "hour_of_day"
    )
    .agg(
        F.count("*").alias("calls"),
        F.sum(
            F.when(F.col("is_dropped") == True, 1).otherwise(0)
        ).alias("dropped_calls"),
        F.avg("duration_seconds_int").alias("avg_duration_seconds")
    )
)

display(calls_hourly.limit(20))
```

```python
calls_hourly = (
    calls_hourly
    .withColumn(
        "drop_rate",
        F.when(
            F.col("calls") > 0,
            F.col("dropped_calls") / F.col("calls")
        ).otherwise(F.lit(0.0))
    )
)

display(
    calls_hourly.orderBy(
        F.col("drop_rate").desc()
    ).limit(20)
)
```

```python
events = (
    silver
    .withColumn("call_date", F.to_date("start_ts_timestamp"))
    .withColumn("hour_of_day", F.hour("start_ts_timestamp"))
    .select(
        "tower_id",
        "call_date",
        "hour_of_day",
        "start_ts_timestamp",
        "end_ts",
        "call_id"
    )
)

start_events = (
    events
    .select(
        "tower_id",
        "call_date",
        "hour_of_day",
        F.col("start_ts_timestamp").alias("event_ts"),
        F.lit(1).alias("delta")
    )
)

end_events = (
    events
    .select(
        "tower_id",
        "call_date",
        "hour_of_day",
        F.col("end_ts").alias("event_ts"),
        F.lit(-1).alias("delta")
    )
)

concurrency_events = start_events.unionByName(end_events)

display(concurrency_events.limit(20))
```

```python
event_points = (
    concurrency_events
    .groupBy(
        "tower_id",
        "call_date",
        "hour_of_day",
        "event_ts"
    )
    .agg(
        F.sum("delta").alias("delta")
    )
)

display(event_points.limit(20))
```

```python
concurrency_window = (
    Window
    .partitionBy(
        "tower_id",
        "call_date",
        "hour_of_day"
    )
    .orderBy("event_ts")
    .rowsBetween(
        Window.unboundedPreceding,
        Window.currentRow
    )
)

concurrency_running = (
    event_points
    .withColumn(
        "concurrent_calls",
        F.sum("delta").over(concurrency_window)
    )
)

display(
    concurrency_running.limit(30)
)
```

```python
peak_concurrency = (
    concurrency_running
    .groupBy(
        "tower_id",
        "call_date",
        "hour_of_day"
    )
    .agg(
        F.max("concurrent_calls").alias("peak_concurrent")
    )
)

display(
    peak_concurrency.orderBy(
        F.col("peak_concurrent").desc()
    ).limit(20)
)
```

```python
gold = (
    calls_hourly
    .join(
        peak_concurrency,
        on=[
            "tower_id",
            "call_date",
            "hour_of_day"
        ],
        how="left"
    )
    .withColumn(
        "utilisation",
        F.col("peak_concurrent") /
        F.col("capacity_channels")
    )
    .select(
        "tower_id",
        "site_name",
        "city",
        "district",
        "capacity_channels",
        "call_date",
        "hour_of_day",
        "calls",
        "dropped_calls",
        "drop_rate",
        "peak_concurrent",
        "utilisation",
        "avg_duration_seconds"
    )
)

display(gold.limit(20))
```

```python
(
    gold.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(GOLD_TABLE)
)

print(f"Gold table created: {GOLD_TABLE}")
```

```python
gold_check = spark.table(GOLD_TABLE)

print("Gold rows:", gold_check.count())

print(
    "Total calls:",
    gold_check.agg(F.sum("calls")).collect()[0][0]
)

print(
    "Total dropped calls:",
    gold_check.agg(F.sum("dropped_calls")).collect()[0][0]
)

print(
    "Overall drop rate:",
    gold_check.agg(
        F.sum("dropped_calls") / F.sum("calls")
    ).collect()[0][0]
)
```

```python
tower_summary = (
    gold_check
    .groupBy("tower_id")
    .agg(
        F.sum("calls").alias("calls"),
        F.sum("dropped_calls").alias("dropped_calls")
    )
    .withColumn(
        "drop_rate",
        F.col("dropped_calls") / F.col("calls")
    )
    .orderBy(F.col("drop_rate").desc())
)

display(tower_summary)
```

```python
utilisation_summary = (
    gold_check
    .groupBy("utilisation")
    .agg(
        F.sum("calls").alias("calls"),
        F.sum("dropped_calls").alias("dropped_calls")
    )
    .withColumn(
        "drop_rate",
        F.col("dropped_calls") / F.col("calls")
    )
    .orderBy("utilisation")
)

display(utilisation_summary)
```

```python
from pyspark.sql.window import Window
from pyspark.sql import functions as F

tower_hour = (
    gold_check
    .select(
        "tower_id",
        "call_date",
        "hour_of_day",
        "calls",
        "dropped_calls",
        "drop_rate"
    )
)

worst_hour_window = (
    Window
    .partitionBy("tower_id")
    .orderBy(
        F.col("drop_rate").desc(),
        F.col("hour_of_day").desc()
    )
)

worst_hour = (
    tower_hour
    .withColumn(
        "rank",
        F.row_number().over(worst_hour_window)
    )
    .filter(F.col("rank") == 1)
    .drop("rank")
    .orderBy("tower_id")
)

display(worst_hour)
```

```python
daily_tower = (
    gold_check
    .groupBy(
        "tower_id",
        "call_date"
    )
    .agg(
        F.sum("calls").alias("calls"),
        F.sum("dropped_calls").alias("dropped_calls")
    )
    .withColumn(
        "drop_rate",
        F.col("dropped_calls") / F.col("calls")
    )
)

towers_over_5pct = (
    daily_tower
    .filter(F.col("drop_rate") > 0.05)
    .groupBy("tower_id")
    .agg(
        F.count("*").alias("days_over_5pct")
    )
    .filter(F.col("days_over_5pct") > 10)
    .orderBy(
        F.col("days_over_5pct").desc(),
        "tower_id"
    )
)

display(towers_over_5pct)
```

```python
concurrency_analysis = (
    gold_check
    .groupBy("peak_concurrent")
    .agg(
        F.sum("calls").alias("calls"),
        F.sum("dropped_calls").alias("dropped_calls")
    )
    .withColumn(
        "drop_rate",
        F.col("dropped_calls") / F.col("calls")
    )
    .orderBy("peak_concurrent")
)

display(concurrency_analysis)
```

```python
print("========== GOLD FINAL VALIDATION ==========")

print("Gold rows:")
print(gold_check.count())

totals = gold_check.agg(
    F.sum("calls").alias("calls"),
    F.sum("dropped_calls").alias("dropped")
).collect()[0]

print("Total calls:", totals["calls"])
print("Total dropped calls:", totals["dropped"])
print("Overall drop rate:", totals["dropped"] / totals["calls"])

print("\nPeak concurrency values:")
(
    gold_check
    .groupBy("peak_concurrent")
    .count()
    .orderBy("peak_concurrent")
    .show()
)

print("\nWorst towers:")
tower_summary.show(10, truncate=False)
```
